### 5) Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de mesure 12h, 13h, 14h et 15h

#### Méthode 1 — en passant index directement :

```python
temperatures_horaires = pd.Series([22.5, 23.1, 21.8, 24.0], index=["12h", "13h", "14h", "15h"])
temperatures_horaires
```

#### Méthode 2 — à partir d'un dictionnaire :

```python
temperatures_horaires = pd.Series({"12h": 22.5, "13h": 23.1, "14h": 21.8, "15h": 24.0})
temperatures_horaires
```

### 6) Afficher la 1ère et la dernière valeur 

```python
premiere = temperatures_horaires.iloc[0]
derniere = temperatures_horaires.iloc[-1]
print(premiere, derniere)
```
